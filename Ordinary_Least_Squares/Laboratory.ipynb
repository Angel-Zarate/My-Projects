## Basic Equations to Calculate Metrics for the Model
The OLS general model with two parameters $\hat{y}$ is: 
$$\hat{y_i}= \theta_0 + \theta_1x_i$$

The error $E$ for one instance:
$$E = y_i - \hat{y}_i$$
But it is squared to remove negative signs as shown:
$$SE = (y_i - \hat{y}_i)^{2}$$
And then, it is applied the sum of all squared errors to calculate for all instances:
$$SSE = \sum_{i=1}^{N}(y_i - \hat{y_i})^2$$

Applying the partial derivative with respect $\theta_0$ and equaliting to zero:
$$\frac{\partial SSE}{\partial \theta_0} = 0$$
Thus
$$\frac{\partial(\sum_{i=1}^{N}(y_i - (\theta_0 + \theta_1x_i))^2)}{\partial \theta_0}=0$$

Also applying partial derivative with respect $\theta_1$ and equaliting to zero:
$$\frac{\partial SSE}{\partial \theta_1} = 0$$
and then;
$$\frac{\partial(\sum_{i=1}^{N}(y_i - (\theta_0 + \theta_1x_i))^2)}{\partial \theta_1}=0$$

And the result of each derivation $\theta_0$ and $\theta_1$ are respectively:
$$\hat{\theta_0} = \bar{y} - \hat{\theta_1} \bar{x}$$
$$\hat{\theta_1} = \frac{\sum_{i=1}^{N}x_i(\bar{y}-y_i)}{\sum_{i=1}^{N}x_i(\bar{x}-x_i)}

# Envolving all the functions to use, so it will be easy calculate for any dataset

```python
def param(x,y):
    # Calculating x and y mean respectively
    xMean = np.mean(x) 
    yMean = np.mean(y)
    

    num = 0
    den = 0
    # Solving for theta 1
    for i in range(len(x)):
        num = num + x[i]*(y[i]-yMean)
        den = den + x[i]*(x[i]-xMean)
    thetaH1 = num/den

    # Solving for theta 0 
    thetaH0 = yMean - thetaH1*xMean

    return thetaH0,thetaH1

def ols(x,parameters):
        # Calculating y hat
        return parameters[0] + parameters[1]*x

def sse(yHat,y):
    # Calculating SSE
    sum = 0
    for i in range(len(y)):
        sum = sum + (y[i] - yHat[i])**2
    return sum
```

# Dataset 1

```python
# Importing libraries to work
import pandas as pd
import numpy as np

data = pd.read_csv("dataset-1.csv")

# Getting the dataset into an array
x = data['x'].to_numpy()
y = data['y'].to_numpy()
N1 = len(x) # number of points

results_1 = param(x,y)
yHat = ols(x,results_1)
SSE_1 = sse(yHat,y)
print(f'SSE = {SSE_1}')
print(f'Points = {N1}')
```

```python
import matplotlib.pyplot as plt

plt.plot(x,y,'*b',label='Original')
plt.plot(x,yHat,'-r',label='Model')
plt.legend()
plt.show()
```
